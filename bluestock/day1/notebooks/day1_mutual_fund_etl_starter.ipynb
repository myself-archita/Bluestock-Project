Place your CSV files in `data/raw/` before running the cells below.

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW_DIR = BASE_DIR / 'data' / 'raw'
RAW_DIR
```

```python
csv_files = sorted(RAW_DIR.glob('*.csv'))
[(csv.name, csv.stat().st_size) for csv in csv_files]
```

```python
def read_csv_file(csv_path):
    for encoding in ('utf-8-sig', 'utf-8', 'latin1'):
        try:
            return pd.read_csv(csv_path, encoding=encoding)
        except UnicodeDecodeError:
            continue
    return pd.read_csv(csv_path, encoding='cp1252')

for csv_file in csv_files:
    frame = read_csv_file(csv_file)
    display(frame.head())
    print(csv_file.name, frame.shape)
    display(frame.dtypes.to_frame('dtype'))
```
