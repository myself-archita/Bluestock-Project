## Setup

Load the cleaned NAV data from `data/processed/`. If the full dataset is not available, you can use the synthetic sample files generated for review.

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path('..').resolve()
PROCESSED_DIR = BASE_DIR / 'data' / 'processed'
nav_path = PROCESSED_DIR / 'nav_history_cleaned.csv'
nav = pd.read_csv(nav_path)
nav['date'] = pd.to_datetime(nav['date'])
nav.head()
```

## Daily Returns

Daily return is computed as `nav_t / nav_t-1 - 1` for each fund. The distribution should be centered near zero, with moderate tails depending on the scheme profile.

```python
nav['daily_return'] = nav.groupby('amfi_code')['nav'].pct_change()
nav['daily_return'].describe()
```
